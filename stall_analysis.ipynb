**Project:** Wanjiku's Fresh produce stall

**Objective:** To analyse 14 days of sales data, evaluate business performace, identity restocking needs and provide actionable business recommendations.


```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall sales perormance

I will use python's built-in sum(), len(), max(), min(), and index() functions to calculate total, average, highest and lowest sales. I will also identify the say and week when the highest and lowest sales occured.

```python
total_sales = sum(daily_sales)
average_sales = total_sales/len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]

best_week = (best_index //7) + 1
worst_week = (worst_index //7) + 1

print("Overall sales performance")
print("-----------------------")
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f} ")
print(f"Best day: {best_day}, Week {best_week}")
print(f"Lowest sales: KES {lowest_sales:,.2f}")
print(f"worst day: {worst_day}, Week {worst_week}")
```

## Task2: Weekly sales and growth

I will create a reusable week_totals() function that uses the list slicing to calculate the total sales for each week. I will then calculate the percentage growth or decline between the two weeks.

```python
def week_totals(sales):
  """Return the total sales for the week one and week tw0."""
  week_one = sum(sales[:7])
  week_two = sum(sales[:7])

  return week_one, week_two
week_one_total, week_two_total = week_totals(daily_sales)

growth = ((week_two_total - week_one_total) / week_one_total) *100

print("Weekly Sales")
print(f"week 1: KES {week_one_total:,.2f}")
print(f"week 2: KES {week_two_total:,.2f}")
print(f"week-on-week growth: {growth:2f}%")


```

## Task 3: Daily Sales Classification

I will create a classify_day() function to classify sales as Strong, Normal or Weak based on the overall daily average. I will use a dictionary to count the number of days in each category.

```python
def classify_day(amount, average):
  """classify daily sales based on 20% above or below the average"""
  if amount > average *  1.20:
    return "strong"
  elif amount < average * 0.80:
    return "weak"
  else:
    return "Normal"

sales_categories = {
    "strong": 0,
    "Normal": 0,
    "weak": 0,
}

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    sales_categories[category] += 1

print("DAILY SALES CLASSIFICATION")


for category, count in sales_categories.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday vs Weekend Sales

I will separate weekday sales (Monday to Friday) from weekend sales (Saturday and Sunday) using list indexing. I will calculate the average sales for both groups and compare their performance using a percentage difference.

```python
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

weekend_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("WEEKDAY VS WEEKEND")
print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend performance difference: {weekend_difference:.2f}%")
```

## Task 5: Stock Reordering and Gross Margins

I will use the stock dictionary and the .items() method to identify products whose quantities are at or below their reorder levels. I will also calculate the gross margin percentage for each product using its selling price and cost price.

```python
def products_to_reorder(stock_data):
    """Return products with stock quantities at or below reorder levels."""
    reorder_list = []

    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_products = products_to_reorder(stock)

print("STOCK AND GROSS MARGINS")

for product, details in stock.items():
    margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    print(f"{product}: Gross margin = {margin:.2f}%")

print("\nPRODUCTS TO REORDER:")

for product in reorder_products:
    print(product)
```

## Task 6: Stall Performance Report

I will combine the results from the previous tasks into one professional report using f-strings. I will format all monetary values in Kenya Shillings with thousands separators for easy reading.

```python
print("=" * 55)
print("          WANJIKU'S STALL PERFORMANCE REPORT")
print("=" * 55)

print("\n1. SALES SUMMARY")
print("-" * 55)
print(f"Total sales:          KES {total_sales:,.2f}")
print(f"Average daily sales:  KES {average_sales:,.2f}")
print(f"Highest daily sales:  KES {highest_sales:,.2f}")
print(f"Lowest daily sales:   KES {lowest_sales:,.2f}")
print(f"Best sales day:       {best_day}, Week {best_week}")
print(f"Worst sales day:      {worst_day}, Week {worst_week}")

print("\n2. WEEKLY PERFORMANCE")
print("-" * 55)
print(f"Week 1 sales:         KES {week_one_total:,.2f}")
print(f"Week 2 sales:         KES {week_two_total:,.2f}")
print(f"Weekly growth:        {growth:.2f}%")

print("\n3. SALES CLASSIFICATION")
print("-" * 55)

for category, count in sales_categories.items():
    print(f"{category:<15} {count} days")

print("\n4. WEEKDAY AND WEEKEND PERFORMANCE")
print("-" * 55)
print(f"Weekday average:      KES {weekday_average:,.2f}")
print(f"Weekend average:      KES {weekend_average:,.2f}")
print(f"Weekend difference:   {weekend_difference:.2f}%")

print("\n5. STOCK AND GROSS MARGINS")
print("-" * 55)

for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) /
              details["price"]) * 100

    print(f"{product:<15} Margin: {margin:>6.2f}%")

print("\nProducts requiring restocking:")

for product in reorder_products:
    print(f"- {product}")

print("\n" + "=" * 55)
print("                 END OF REPORT")
print("=" * 55)
```
